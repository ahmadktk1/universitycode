### 1. Introduction to Python

### 2. What is a Variable?
A variable is used to store data in memory.
Syntax:
```
variable_name = value
```


````python

age = 25
name = "Ali"
height = 5.8
print(age)
print(name)
print(height)
````


### 3. Rules for Naming Variables
- Must start with a letter or underscore
- Cannot start with a number
- Case-sensitive


````python

student_name = "Ayesha"
_student_id = 101
print(student_name, _student_id)
````


### 4. Data Types
Python automatically detects data types.


````python

marks = 90
percentage = 87.5
university = "SS CASE IT"
is_passed = True

print(type(marks))
print(type(percentage))
print(type(university))
print(type(is_passed))
````


### 5. print() Function
Used to display output on the screen.


````python

print("Hello, Python!")
print("Name:", name)
print(f"My name is {name} and I am {age} years old.")  #An f-string (formatted string) is a clean and easy way to print variables inside text.
````
