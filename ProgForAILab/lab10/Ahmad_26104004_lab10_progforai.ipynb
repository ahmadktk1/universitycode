5. Given two arrays a and b, find the common elements between them.

```python
np.random.seed(34)
a = np.random.randint(1,20,size=(3,3))
b = np.random.randint(10,30,size=(3,3))

common = np.intersect1d(a, b)
print(f"array a \n {a}")

print(f"array b\n {b}")

print(f"\n common in a | b : {common}")


```
