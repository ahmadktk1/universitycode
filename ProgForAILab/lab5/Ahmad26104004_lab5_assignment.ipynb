**Task 1:**

Library Catalog System
Build a library catalog system that keeps track of books, authors, and their availability using
dictionaries, lists, and tuples.
Create a dictionary where the keys are authors&#39; names, and the values are lists of tuples. Each
tuple represents a book with details (e.g., title, publication year, availability status).
Implement functions to:
- Add new books to the catalog.
- Mark a book as borrowed or returned.
- Display a list of books by a specific author.
- Find and display the oldest and newest books in the catalog.
- List books that are currently available for borrowing.

```python

catalog = {}


def add_book(author, title, year):
    book = (title, year, True)  
    
    if author in catalog:
        catalog[author].append(book)
    else:
        catalog[author] = [book]
    
    print(f'"{title}" added to the catalog.')

# function for book borrowing
def borrow_book(title):
    for author in catalog:
        for i, book in enumerate(catalog[author]):
            if book[0] == title:
                if book[2]:
                    catalog[author][i] = (book[0], book[1], False)
                    print(f'You borrowed "{title}".')
                else:
                    print(f'"{title}" is already borrowed.')
                return
    print("Book not found.")

# return  a book
def return_book(title):
    for author in catalog:
        for i, book in enumerate(catalog[author]):
            if book[0] == title:
                catalog[author][i] = (book[0], book[1], True)
                print(f'"{title}" has been returned.')
                return
    print("Book not found.")

# display the book by author
def display_books_by_author(author):
    if author in catalog:
        print(f"\nBooks by {author}:")
        for book in catalog[author]:
            status = "Available" if book[2] else "Borrowed"
            print(f"{book[0]} ({book[1]}) - {status}")
    else:
        print("Author not found.")

# Find oldest and newest books
def oldest_newest_books():
    all_books = []
    
    for books in catalog.values():
        all_books.extend(books)
    
    if not all_books:
        print("No books in catalog.")
        return
    
    oldest = min(all_books, key=lambda x: x[1])
    newest = max(all_books, key=lambda x: x[1])
    
    print(f"Oldest book: {oldest[0]} ({oldest[1]})")
    print(f"Newest book: {newest[0]} ({newest[1]})")

# List available books
def available_books():
    print("\nAvailable Books:")
    found = False
    
    for author, books in catalog.items():
        for book in books:
            if book[2]:
                print(f"{book[0]} by {author} ({book[1]})")
                found = True
    
    if not found:
        print("No books available.")
```

```python
add_book("J.K. Rowling", "Harry Potter and the Sorcerer's Stone", 1997)
add_book("J.K. Rowling", "Harry Potter and the Chamber of Secrets", 1998)
add_book("George Orwell", "1984", 1949)

display_books_by_author("J.K. Rowling")

borrow_book("1984")
available_books()

oldest_newest_books()

return_book("1984")
available_books()
```

**Task 5:**

Write a program that finds the absolute difference between the sums of its diagonals.

```python
# Matrix
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

n = len(matrix)
d1 = 0
d2 = 0

for i in range(n):
    d1 += matrix[i][i]          # main diagonal
    d2 += matrix[i][n-i-1]      # secondary diagonal

difference = abs(d1 - d2)

print("Absolute Difference:", difference)
```

**Task 7:**

Write a program that check a list contains a palindrome of elements (use copy() method)
Example
[1,2,3,2,1]
[1, “abc”, “abc”, 1]

```python
lst = [1, 2, 3, 2, 1]
lst2 = [1, 'abc', 'abc', 1]
# Copy the list
rev = lst.copy()
rev2 = lst2.copy()
# Reverse the copied list
rev.reverse()
rev2.reverse()

# Check palindrome
if lst  == rev and lst2 == rev2:
    
    print(f"Reversed List : {rev}")
    print(f"reversed list2 : {rev2}")
    print("both list are Palindrome")
else:
    print("List is Not Palindrome")
```

**Task 10:**

You are given the following tuple:
data = (10, 20, 30, 40, 50, 60, 70, 80)
Perform the following operations without using loops, functions, or methods:
 Extract the first three elements and the last three elements, then combine them into a new
tuple.
Expected: (10, 20, 30, 60, 70, 80)
 Reverse the tuple using slicing.
Expected: (80, 70, 60, 50, 40, 30, 20, 10)
 Create a tuple with only the even-indexed elements.
Expected: (10, 30, 50, 70)
 Create a tuple with only the odd-indexed elements.
Expected: (20, 40, 60, 80)
 Swap the first half with the second half of the tuple.
Expected: (50, 60, 70, 80, 10, 20, 30, 40)

```python
data = (10, 20, 30, 40, 50, 60, 70, 80)

#  First three + last three
new_tuple = data[:3] + data[-3:]
print(new_tuple)

#  Reversing the tuple
reverse_tuple = data[::-1]
print(reverse_tuple)

# Even-indexed elements
even_index = data[::2]
print(even_index)

# Odd-indexed elements
odd_index = data[1::2]
print(odd_index)

# Swapping first half with second half
swap = data[4:] + data[:4]
print(swap)
```
