**What is Type Casting in Python?**

Type casting (or type conversion) is the process of changing one data type into another.

For example:

  * Converting a string "10" into an integer 10.

  * Converting a float 12.5 into an integer 12.

Python has built-in functions for type casting:

  * int() → converts to integer

  * float() → converts to float

  * str() → converts to string

**String → Integer**

```python
num_str = "50"      # string
num_int = int(num_str)
print(num_int, type(num_int))
```

**String → Float**

```python
price_str = "12.75"
price_float = float(price_str)
print(price_float, type(price_float))
```

**Float → Integer**

```python
num_float = 45.9
num_int = int(num_float)
print(num_int, type(num_int))
```

**Integer → String**

```python
age = 20
age_str = str(age)
print(age_str, type(age_str))
```

**Upcasting and Downcasting in Type Casting**

* Upcasting → Converting a smaller data type into a larger/more general one (safe conversion).

* Downcasting → Converting a larger/more precise data type into a smaller/less precise one (may lose information).

```python
# Upcasting (int → float)
a = 10      # integer
b = float(a)
print(b, type(b))

# Downcasting (float → int)
x = 15.75
y = int(x)
print(y, type(y))
```

## Lists
A list is a collection of values enclosed in square brackets []. It can hold elements of different types.

### List Characteristics:

* Lists are ordered collections.
* Lists are mutable (can be changed after creation).
* Lists can contain duplicate elements.

```python
my_list = [1, 2, 3, 'apple', 'mango'] # creating a list

print(my_list) #print a list
print(my_list[3]) # print values from index 3
print(my_list[2:5]) # print values from index 2 to 5 (5 exclusive)
```

**List Operations**

```python
a = [1, 2, 3]
b = [4, 5, 6]

# Concatenation (+)
print("Concatenation:", a + b)

# Repetition (*)
print("Repetition:", a * 2)

# Membership (in / not in)
print("Membership:", 2 in a)
print("Not in:", 10 not in a)
```

**List Slicing**

```python
nums = [10, 20, 30, 40, 50, 60]

# Positive Indexing
print(nums[0])
print(nums[2:5])
print(nums[:3])
print(nums[3:])
```

```python
nums= [10, 20, 30, 40, 50, 60]
# Negative Indexing
print(nums[-1])
print(nums[-3:-1])
```

```python
# With Step
nums= [10, 20, 30, 40, 50, 60]
print(nums[::2])
print(nums[1:5:2])
print(nums[::-1])
print(nums[-1:-6:-2])
```

Further operations on lists are here:

* append()   : Used for appending and adding elements to the end of the List.

* copy()     : It returns a copy of a list.        

* clear()    : This method is used for removing all items from the list.   

* count()    : These methods count the elements.                            

* extend()   : Adds each element of the iterable to the end of the List.     

* index()    : Returns the lowest index where the element appears.           

* insert()   : Inserts a given element at a given index in a list.          

* pop()      : Removes and returns the last value from the List or the given index value.

* remove()   : Removes a given object from the List.                        

* reverse()  : Reverses objects of the List in place.                       

* sort()     : Sort a List in ascending, descending, or user-defined order.

* min()      : Calculates the minimum of all the elements of the List.      

* max()      : Calculates the maximum of all the elements of the List.      

```python
fruits = ["apple", "dates", "cherry"]

# Adding elements
fruits.append("mango")       # add at end
print("Append:", fruits)

fruits.insert(1, "orange")   # add at specific index
print("Insert:", fruits)

fruits.extend(["grape", "melon"])  # extend list
print("Extend:", fruits)

# Removing elements
fruits.remove("dates")   # removes first occurrence
print("Remove:", fruits)

last = fruits.pop()       # removes last element
print("Pop:", fruits, "Popped:", last)

del fruits[0]             # delete by index
print("Del:", fruits)

fruits.clear()            # remove all elements
print("Clear:", fruits)


```

```python
nums = [10, 20, 30, 40, 10]

# Searching
print("Index of 30:", nums.index(30))
print("Count of 10:", nums.count(10))

# Sorting
nums.sort()
print("Sort:", nums)

nums.sort(reverse=True)
print("Reverse Sort:", nums)

# Copying
nums_copy = nums.copy()
print("Copy:", nums_copy)

# Other built-ins
print("Length:", len(nums))
print("Max:", max(nums))
print("Min:", min(nums))
print("Sum:", sum(nums))
```

# **2d list**

A 2D list is simply a list where each element itself is a list.

```python
# Creating a 2D list (3x3 matrix)
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print("Matrix:")
for row in matrix:
    print(row)
```

# **Dictionary**

Key-value pairs {key:value}

```python
# dictionary
student = {
    "name":"Ali",
    "age":13,
    "grade": "A"
}

print(student)
print(student["grade"])
```

# **Operations on Dictionaries**

we can merge dictionaries using update() or |

```python
d1 = {"a": 1, "b": 2}
d2 = {"a": 3, "d": 4}

# Using update
d1.update(d2)
print("Merged with update:", d1)


```

```python
# Using |
d3 = {"x": 10} | {"y": 20}
print("Merged with | :", d3)
```

**Membership**

* Membership works only on keys, not values.

```python
d = {"name": "Ali", "age": 30, "city": "Islamabad"}

print("name" in d)
print("Ali" in d)
print("country" not in d)
```

# **Important Dictionary Functions**

```python
d = {"a": 10, "b": 20, "c": 30}
# Length
print(len(d))

# Access values
print(d["a"])
print(d.get("b"))
print(d.get("x", "Not Found"))  # Safe access

# Add / Update key
d["d"] = 40
d["a"] = 99
print(d)

# Delete key
d.pop("b")
print(d)

# Delete last inserted item
d.popitem()
print(d)

# Keys, Values, Items
print(d.keys())
print(d.values())
print(d.items())

# Clear dictionary
d.clear()
print(d)  # {}
```

```python
# nested dictionary
cars = {
    "brand":{"side":"Toyota", "main":"Tesla"},
    "model_no": 546,
    "year_of_purchase":"2025"
}
print(cars)
print(cars["brand"])
print(cars["brand"]["main"])
```
