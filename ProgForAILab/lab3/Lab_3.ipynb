```python
## Even or Odd Number
num = int(input("Enter a number: "))

if num % 2 == 0:
    print(num, "is even")
    print("It is divisible by 2")
    print("End of even check")
else:
    print(num, "is odd")
    print("It is not divisible by 2")
    print("End of odd check")
```

## 4. Nested if

```python
#Even and Positive Number
num = int(input("Enter a number: "))

if num > 0:
    if num % 2 == 0:
        print(num, "is positive and even")
    else:
        print(num, "is positive but odd")
else:
    print(num, "is not positive")
```

```python
#Voting Eligibility + Nationality
age = int(input("Enter your age: "))
citizen = input("Are you a citizen? (yes/no): ")

if age >= 18:
    if citizen.lower() == "yes":
        print("You are eligible to vote.")
    else:
        print("You must be a citizen to vote.")
else:
    print("You are too young to vote.")
```

```python
##ATM Simulation

balance = 5000
pin = int(input("Enter your PIN: "))

if pin == 1234:
    print("PIN accepted.")
    withdraw = int(input("Enter withdrawal amount: "))
    if withdraw <= balance:
        if withdraw % 500 == 0:
            balance -= withdraw
            print("Transaction successful.")
            print("Remaining balance:", balance)
        else:
            print("Amount must be in multiples of 500.")
    else:
        print("Insufficient balance.")
else:
    print("Invalid PIN.")
```

```python
#Student Result with Nested Conditions
marks = int(input("Enter your marks: "))

if marks >= 40:
    print("Pass")
    if marks >= 60:
        if marks >= 80:
            print("Excellent performance!")
        else:
            print("Good performance!")
    else:
        print("Satisfactory performance.")
else:
    print("Fail")
```
