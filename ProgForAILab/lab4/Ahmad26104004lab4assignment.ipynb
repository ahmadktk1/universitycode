**2. Write a program using a for loop that sums only odd numbers between 1 and 20 —
but you cannot use an if statement inside the loop.**

```python
odd_sum = 0
for num in range(1,20,2):
    odd_sum += num

print(f"Sum of odd numbers between 1 to 20 : {odd_sum}")
```

**3. Continuously take integer input from the user and find their square. Stop when the user enters a negative number.**

```python
while True:
    user_input = int(input("Number : "))
    if user_input < 0:
        print("Positives only")
        break
    sqaured = user_input ** 2
    print(f"Square of {user_input} is : {sqaured}")
```

**4. Count how many digits are in a given number using a while loop (no string conversion).**

```python
number = int(input("Number : "))
temp_no = number
count = 0

# the below line will make the numbers only positives
number = abs(number)
while number > 0:
    if number == 0:
        print(f"Digits counts {number}")

    number //= 10 #This will do integer division and save it to number
    count += 1 # this will count the digits (every time we do integer division it increament 1)

print(f"Total digits in {temp_no} is : {count}")
```

**5.Reverse an integer using a while loop. (Example: Input = 123 → Output = 321)**

```python
num = 123
reversed_num = 0
# Reverse the number using while loop
while num > 0:
    digit = num % 10           # Get last digit
    reversed_num = reversed_num * 10 + digit  # Add digit to reversed number
    num = num // 10            # Remove last digit

# Handle negative numbers
if num < 0:
    reversed_num = -reversed_num

# Print the result
print(f"Reversed number: {reversed_num}")
```

**7. Write a code to print following number using break and continue statement:
1, 2, 3, 5, 6, 7, 9**

This code will have a range from 1 to 20 but the task says use continue and break statements so as per question the loop will continue on 4 and 8 and will break when reach 10

```python
# This code will have a range from 1 to 20 but 
for number in range(1,20):
    if number == 4 and number == 8:
        continue
    elif number == 10:
        break
    print(number)
```
