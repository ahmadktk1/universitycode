## 1. while Loop

```python
i = 1
while i <= 5:
    print(i)
    i += 1
```

```python
n = 5
i = 1
sum1 = 0
while i <= n:
    sum1 += i
    i += 1
print("Sum =", sum1)
```

```python
##Multiplication Table

num = 7
i = 1

while i <= 10:
    print(num, "x", i, "=", num * i)
    i += 1
```

```python
#Reverse Countdown

i = 10
while i >= 1:
    print(i)
    i -= 1
```

```python
##Guess the Number Game

secret = 7
guess = 0

while guess != secret:
    guess = int(input("Guess the number (1-10): "))
    if guess == secret:
        print("Correct! You guessed it.")
```

```python
##ATM Simulation

balance = 5000
pin = int(input("Enter PIN: "))

while pin != 1234:
    print("Incorrect PIN, try again.")
    pin = int(input("Enter PIN: "))

print("Login successful!")
withdraw = int(input("Enter amount to withdraw: "))

while withdraw > balance:
    print("Insufficient balance. Try smaller amount.")
    withdraw = int(input("Enter amount to withdraw: "))

balance -= withdraw
print("Transaction successful. Remaining balance:", balance)
```

```python
#Factorial Calculation

num = int(input("Enter a number: "))
fact = 1
i = 1

while i <= num:
    fact *= i
    i += 1

print("Factorial of", num, "is", fact)
```

```python
#Increasing Stars
i = 1
while i <= 5:
    print("*" * i)
    i += 1
```

```python
#Decreasing Stars
i = 5
while i >= 1:
    print("*" * i)
    i -= 1
```

```python
#Right-Aligned Triangle
i = 1
while i <= 5:
    print(" " * (5 - i) + "*" * i)
    i += 1
```

```python
#Pyramid Pattern
i = 1
while i <= 5:
    print(" " * (5 - i) + "*" * (2 * i - 1))
    i += 1
```

```python
#Number Triangle
i = 1
while i <= 5:
    j = 1
    while j <= i:
        print(j, end=" ")
        j += 1
    print()
    i += 1
```

```python
#Inverted Number Triangle
i = 5
while i >= 1:
    j = 1
    while j <= i:
        print(j, end=" ")
        j += 1
    print()
    i -= 1
```

## 2. Infinite Loop

```python
while True:
    print("This will run forever!")
```

```python
while True:
    user_input = input("Enter 'q' to quit: ")
    if user_input == 'q':
        break
    print("You typed:", user_input)
```

## 3. for Loop

```python
for ch in "Python":
    print(ch)
```

```python
fruits = ["apple", "banana", "cherry"]
for fruit in fruits:
    print(fruit)
```

```python
#Iterate Over a Tuple
colors = ("red", "green", "blue")

for color in colors:
    print("Color:", color)
```

```python
#Iterate Over a Set
numbers = {10, 20, 30, 40}

for num in numbers:
    print("Number:", num)
```

```python
num = 5
fact = 1
for i in range(1, num + 1):
    fact *= i
print("Factorial:", fact)
```

```python
student = {"name": "Ali", "age": 21, "grade": "A"}

for key, value in student.items():
    print(key, ":", value)
```

```python
for i in range(5):
    print(i)
```

```python
#Print Numbers from 1 to 10
for i in range(1, 11):
    print(i)
```

```python
#Print Even Numbers
for i in range(2, 21, 2):
    print(i)
```

```python
#Sum of Numbers
total = 0
for i in range(1, 6):
    total += i
print("Sum =", total)
```

```python
#Multiplication Table
num = 7
for i in range(1, 11):
    print(num, "x", i, "=", num * i)
```

```python
#Factorial Calculation
num = 5
fact = 1
for i in range(1, num + 1):
    fact *= i
print("Factorial of", num, "is", fact)
```

## 4. break Statement

```python
for i in range(1, 10):
    if i == 5:
        break
    print(i)
```

## 5. continue Statement

```python
for i in range(1, 10):
    if i == 5:
        continue
    print(i)
```

## 6. pass Statement

```python
for i in range(5):
    if i == 3:
        pass  # placeholder, does nothing
    else:
        print(i)
```

## 7. range() Function

```python
for i in range(5):
    print(i)
```

```python
for i in range(2, 20, 3):
    print(i)
```

```python
for i in range(10, 0, -2):
    print(i)
```

## 8 Nested loop

```python
#Nested for Loop (Pattern)
for i in range(1, 6):
    for j in range(1, i + 1):
        print("*", end="")
    print()
```

```python
#Print a Rectangle of Stars
i = 1
while i <= 3:          # outer loop → rows
    j = 1
    while j <= 5:      # inner loop → columns
        print("*", end=" ")
        j += 1
    print()
    i += 1
```

```python
#Number Grid
i = 1
while i <= 3:
    j = 1
    while j <= 3:
        print(j, end=" ")
        j += 1
    print()
    i += 1
```

```python
#Multiplication Table (1–5)
i = 1
while i <= 5:
    j = 1
    while j <= 5:
        print(i * j, end="\t")
        j += 1
    print()
    i += 1
```

```python
#Right-Angled Triangle of Stars
i = 1
while i <= 5:
    j = 1
    while j <= i:
        print("*", end="")
        j += 1
    print()
    i += 1
```

```python
#Inverted Number Pyramid
i = 5
while i >= 1:
    j = 1
    while j <= i:
        print(j, end=" ")
        j += 1
    print()
    i -= 1
```

```python
#Triangle
i = 1
num = 1
while i <= 5:
    j = 1
    while j <= i:
        print(num, end=" ")
        num += 1
        j += 1
    print()
    i += 1
```
