```python
cipher = "dvvkzecfssprkkve"

for k in range(26):
    decrypt = ""
    for char in cipher:
        if char.isalpha():
            decrypt += chr(((ord(char.upper()) +65 - k) % 26) + 65)
        else:
            decrypt += char  #
    print(f"{k} : {decrypt}")
```
