```python
def playfair_encrypt(text, key):
    key = key.upper().replace("J","I")
    alphabet = "ABCDEFGHIKLMNOPQRSTUVWXYZ"
    
    # build matrix
    matrix = []
    for c in key + alphabet:
        if c not in matrix:
            matrix.append(c)
    matrix = [matrix[i:i+5] for i in range(0,25,5)]

    # prepare text
    text = text.upper().replace("J","I").replace(" ","")
    pairs = []
    i = 0
    while i < len(text):
        a = text[i]
        b = text[i+1] if i+1 < len(text) else "X"
        if a == b:
            pairs.append(a+"X")
            i += 1
        else:
            pairs.append(a+b)
            i += 2

    # encryption
    result = ""
    for a,b in pairs:
        r1,c1 = [(r,c) for r in range(5) for c in range(5) if matrix[r][c]==a][0]
        r2,c2 = [(r,c) for r in range(5) for c in range(5) if matrix[r][c]==b][0]

        if r1 == r2:
            result += matrix[r1][(c1+1)%5] + matrix[r2][(c2+1)%5]
        elif c1 == c2:
            result += matrix[(r1+1)%5][c1] + matrix[(r2+1)%5][c2]
        else:
            result += matrix[r1][c2] + matrix[r2][c1]

    return result


print(playfair_encrypt("INSTRUMENTS","MONARCHY"))
```
