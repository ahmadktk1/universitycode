Tasks:
1. Perform brute-force decryption by trying all possible shifts (0–25).
2. Identify the meaningful plaintext.
3. Verify your result with frequency analysis (most frequent ciphertext letter vs. expected English
distribution)

```python
ciphertext = "dvvkzecfssprkkve"
# formula for decryption of ceasor cihper = C(p,k) = (p-k)mod 26

for k in range(0,26):
    decrypt = ""
    for char in ciphertext:
        if char.isalpha():
            decrypt += chr((ord(char.upper())+65 - k) % 26+65)
        
    else:
        decrypt += char
    print(f"Shift {k} : {decrypt}")
```

We can see that it has a shift of 17
