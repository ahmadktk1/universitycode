Call this function in the main program and print results.

### Question 2: [Marks 5]

Create a program that:

Continuously asks the user to enter numbers until they type 'stop'.
Uses a while loop to calculate and display the average of all even numbers entered.
If no even number is entered, display an appropriate message.

```python
total = 0
count = 0
while True:
    user_input = input("Enter a number (or type 'stop' to finish): ")
    if user_input == "stop":
        break
    else:
        num = int(user_input)
        
        if num % 2 == 0:
            total += num
            count += 1

if count > 0:
    average = total / count
    print("Average of even numbers:", average)
else:
    print("No even numbers were entered.")
    

    
```

### Question 4: [Marks 5]

Write a Python program that takes a dictionary of students and their marks as input (e.g., {'Ali': 85, 'Sara': 92, 'Bilal': 45}) and:

Uses if-elif-else to categorize each student as ‘Pass with Distinction’, ‘Pass’, or ‘Fail’.
Store the categorized results in a new dictionary.
Finally, display the total count of students in each category.

```python
total_students = int(input("Total Students : "))
students = {}
for i in range(total_students):
    name = input("Name: ")
    marks = int(input("Marks: "))
    students[name]=marks

passed = {}
fail = {}

for a in students.keys():
    if students[a]> 60:
        passed[a] = students[a]
    else:
        fail[a]=students[a]

print(f"Total Stuents: {total_students}")
print(f"All Students:\n{students}")
print(f"Passed: \n {passed}")
print(f"Failed: \n{fail}")
```

### Question 5: [Marks 25]

You are asked to build a mini student management system using Python data structures. Store the following information about students using a list of dictionaries. Each dictionary should contain:

"roll_no" (int)
"name" (string)
"courses" (a tuple of enrolled course names, e.g., ("AI", "DS", "CV"))
"marks" (a dictionary where keys are course names and values are integers representing marks).
Example structure (you must extend it for at least 5 students):

students = [

{

"roll_no": 101,

"name": "Ali",

"courses": ("AI", "ML", "CV"),

"marks": {"AI": 85, "ML": 90, "CV": 78}

},

...

]

Write Python code to perform the following tasks:

- Print the name and roll number of the student with the highest total marks across all courses.
- For each course (AI, ML, CV, etc.), compute the highest scorer and display their name, roll number, and marks.
- Update marks of one student by adding 5 bonus points in all courses where they scored less than 80.
- Display a list of all unique courses offered across all students (must be built from the tuples).
- Find the student(s) who are enrolled in exactly the same tuple of courses (check for equality of tuples).
- Remove a student by roll number, and ensure that your updated list reflects this change.
- You must write the program without using external libraries (only built-in Python).

```python
students = [
    {
        "roll_no": 101,
        "name": "Ali",
        "courses": ("AI", "ML", "CV"),
        "marks": {"AI": 85, "ML": 90, "CV": 78}
    },
    {
        "roll_no": 102,
        "name": "Sara",
        "courses": ("AI", "DS", "CV"),
        "marks": {"AI": 88, "DS": 92, "CV": 81}
    },
    {
        "roll_no": 103,
        "name": "Ahmed",
        "courses": ("ML", "DS", "CV"),
        "marks": {"ML": 75, "DS": 80, "CV": 70}
    },
    {
        "roll_no": 104,
        "name": "Ayesha",
        "courses": ("AI", "ML", "CV"),
        "marks": {"AI": 95, "ML": 85, "CV": 90}
    },
    {
        "roll_no": 105,
        "name": "Bilal",
        "courses": ("AI", "DS"),
        "marks": {"AI": 60, "DS": 75}
    }
]

highest_total = 0
top_student = None

for s in students:
    total = sum(s["marks"].values())
    if total > highest_total:
        highest_total = total
        top_student = s

print("Top Student Overall:")
print(top_student["name"], "-", top_student["roll_no"], "Total Marks:", highest_total)

# 2. Highest scorer in each course
course_toppers = {}

for s in students:
    for course, mark in s["marks"].items():
        if course not in course_toppers or mark > course_toppers[course]["marks"]:
            course_toppers[course] = {
                "name": s["name"],
                "roll_no": s["roll_no"],
                "marks": mark
            }

print("Course Toppers:")
for course, info in course_toppers.items():
    print(course, "->", info)


# 3. Update marks (add 5 bonus where marks < 80) for one student (example: roll_no = 105)
for s in students:
    if s["roll_no"] == 105:
        for course in s["marks"]:
            if s["marks"][course] < 80:
                s["marks"][course] += 5

print("Updated Marks for roll_no 105:")
for s in students:
    if s["roll_no"] == 105:
        print(s)


# 4. List all unique courses
unique_courses = set()

for s in students:
    unique_courses.update(s["courses"])

print("Unique Courses:", unique_courses)


# 5. Students with exactly same course tuples
print("Students with same course combinations:")
for i in range(len(students)):
    for j in range(i + 1, len(students)):
        if students[i]["courses"] == students[j]["courses"]:
            print(students[i]["name"], "and", students[j]["name"],
                  "have same courses:", students[i]["courses"])


# 6. Remove a student by roll number (example: remove roll_no = 103)
roll_to_remove = 103

students = [s for s in students if s["roll_no"] != roll_to_remove]

print("\nUpdated Student List after removal:")
for s in students:
    print(s)
```
